````python
import pandas as pd
import numpy as np

import seaborn as sns
from matplotlib import pyplot as plt
%matplotlib inline
````

````python
path = "car_fuel_efficiency.csv"
df = pd.read_csv(path)
df.sample(3, random_state=1)
````

````python
print(df.isna().sum())
df.info()
````

````python
numerical_features = df.select_dtypes(include=np.number).columns.tolist()
categorical_features = df.select_dtypes(include="object").columns.tolist()

df[numerical_features] = df[numerical_features].fillna(0)
df[categorical_features] = df[categorical_features].fillna("NA")
print(df.isna().sum())
````

````python
df.head()
````

````python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction import DictVectorizer
````

````python
df_full_train, df_test = train_test_split(df, test_size=0.2, random_state=1)
df_train, df_val = train_test_split(df_full_train, test_size=0.25, random_state=1)
````

````python
df_train.reset_index(drop=True, inplace=True)
df_val.reset_index(drop=True, inplace=True)
df_test.reset_index(drop=True, inplace=True)

y_train = df_train.fuel_efficiency_mpg.values
y_val = df_val.fuel_efficiency_mpg.values
y_test = df_test.fuel_efficiency_mpg.values

del df_train["fuel_efficiency_mpg"]
del df_val["fuel_efficiency_mpg"]
del df_test["fuel_efficiency_mpg"]
````

````python
dv = DictVectorizer(sparse=False)
X_train = dv.fit_transform(df_train.to_dict(orient='records'))
X_val = dv.transform(df_val.to_dict(orient='records'))
````

## Question 2

Train a random forest regressor with these parameters:

* `n_estimators=10`
* `random_state=1`
* `n_jobs=-1` (optional - to make training faster)


What's the RMSE of this model on the validation data?

* 0.045

````python
from sklearn.ensemble import RandomForestRegressor
````

````python
rf = RandomForestRegressor(n_estimators=10, random_state=1, n_jobs=-1)
````

````python
y_pred = rf.predict(X_val)
````

````python
def rmse(y, y_pred):
    return np.sqrt(np.mean((y-y_pred)**2))


print(f'{rmse(y_val, y_pred):.5f}')
````

## Question 3

Now let's experiment with the `n_estimators` parameter

* Try different values of this parameter from 10 to 200 with step 10.
* Set `random_state` to `1`.
* Evaluate the model on the validation dataset.


After which value of `n_estimators` does RMSE stop improving?
Consider 3 decimal places for calculating the answer.

- 10
- 25
- 80
- 200

If it doesn't stop improving, use the latest iteration number in
your answer.

````python
scores = []

for n in range(10, 201, 10):
    rf = RandomForestRegressor(n_estimators=n, random_state=1)
    rf.fit(X_train, y_train)
    y_pred = rf.predict(X_val)
    score = rmse(y_val, y_pred)
    scores.append((n, score))
    print(f"n_estimators={n:3d}  RMSE={score:.5f}")
````

## Question 4

Let's select the best `max_depth`:

* Try different values of `max_depth`: `[10, 15, 20, 25]`
* For each of these values,
  * try different values of `n_estimators` from 10 till 200 (with step 10)
  * calculate the mean RMSE
* Fix the random seed: `random_state=1`


What's the best `max_depth`, using the mean RMSE?

* 10
* 15
* 20
* 25

````python
results = []

for d in [10, 15, 20, 25]:
    rmses = []
    for n in range(10, 201, 10):
        rf = RandomForestRegressor(max_depth=d, n_estimators=n, random_state=1)
        rf.fit(X_train, y_train)
        y_pred = rf.predict(X_val)
        rmses.append(rmse(y_val, y_pred))
    mean_rmse = np.mean(rmses)
    results.append((d, mean_rmse))
    print(f"max_depth={d:2d}  mean_RMSE={mean_rmse:.5f}")
````

# Question 5

We can extract feature importance information from tree-based models.

At each step of the decision tree learning algorithm, it finds the best split.
When doing it, we can calculate "gain" - the reduction in impurity before and after the split.
This gain is quite useful in understanding what are the important features for tree-based models.

In Scikit-Learn, tree-based models contain this information in the
[`feature_importances_`](https://scikit-learn.org/stable/modules/generated/sklearn.ensemble.RandomForestRegressor.html#sklearn.ensemble.RandomForestRegressor.feature_importances_)
field.

For this homework question, we'll find the most important feature:

* Train the model with these parameters:
  * `n_estimators=10`,
  * `max_depth=20`,
  * `random_state=1`,
  * `n_jobs=-1` (optional)
* Get the feature importance information from this model


What's the most important feature (among these 4)?

* `vehicle_weight`
*	`horsepower`
* `acceleration`
* `engine_displacement`


````python
rf = RandomForestRegressor(n_estimators=10, max_depth=20, random_state=1, n_jobs=-1)
rf.fit(X_train, y_train)
````

````python
importances = rf.feature_importances_
features = list(dv.get_feature_names_out())
````

````python
feature_imp_df = pd.DataFrame({'Feature': features, 'Feature Importance': importances}).sort_values('Feature Importance', ascending=False)
print(feature_imp_df)
````

Now let's train an XGBoost model! For this question, we'll tune the `eta` parameter:

* Install XGBoost


Now change `eta` from `0.3` to `0.1`.

Which eta leads to the best RMSE score on the validation dataset?

* 0.3
* 0.1
* Both give equal value

````python
import xgboost as xgb
````

````python
features = list(dv.get_feature_names_out())
dtrain = xgb.DMatrix(X_train, label=y_train, feature_names=features)
dval = xgb.DMatrix(X_val, label=y_val, feature_names=features)
````

````python
xgb_params = {
    'eta' : 0.3,
    'max_depth' : 6,
    'min_child,weight' : 1,

    'object' : 'reg:squarederror',
    'nyhread' : 8,

    'seed' : 1,
    'verbosity' : 1,
}

model = xgb.train(xgb_params, dtrain, num_boost_round=200)
````

````python
y_pred = model.predict(dval)
print(f'{rmse(y_val, y_pred):.6f}')
````

````python
xgb_params = {
    'eta' : 0.1,
    'max_depth' : 6,
    'min_child,weight' : 1,

    'object' : 'reg:squarederror',
    'nyhread' : 8,

    'seed' : 1,
    'verbosity' : 0,
}

model = xgb.train(xgb_params, dtrain, num_boost_round=200)

y_pred = model.predict(dval)
print(f'{rmse(y_val, y_pred):.6f}')
````
