#### 1.2. Проектируем архитектуру для этой задачи

```python
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense


model = Sequential([
    Dense(26, input_shape=(13,), activation='relu'),
    Dense(13, activation='relu'),
    Dense(1)
])

model.summary()
```

#### 1.3. Компилируем модель ИНС

Подберите нужную функцию потерь и метрику.

```python
from keras.optimizers import Adam

INIT_LR = 0.0001
EPOCHS = 300

opt = Adam(learning_rate=INIT_LR)
model.compile(
    loss='mse',
    optimizer=opt,
    metrics=['mae']
)
```

#### 1.4. Обучаем модель ИНС

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import StandardScaler

X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=0)
X_train.shape, X_test.shape

scaler = StandardScaler()
# scaler = MinMaxScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

print(X_train.shape, X_test.shape)
```

```python
X_train_scaled = pd.DataFrame(X_train, columns=x.columns)  # если x был DataFrame
print(X_train_scaled.describe())
print(x.describe())
```

```python
history = model.fit(
    X_train, y_train,
    epochs=EPOCHS,
    batch_size=16,
    validation_split=0.2,
    verbose=1
)
```

#### 1.6. Делаем предсказания на тестовых данных

```python
import pandas as pd

y_pred = model.predict(X_test).flatten()

results = pd.DataFrame({
    "y_true": y_test.values if hasattr(y_test, "values") else y_test,
    "y_pred": y_pred
})

print(results.head(5))
```

#### 1.7. Вычисляем метрику качества на тестовых данных

```python
test_loss, test_mae = model.evaluate(X_test, y_test, verbose=0)
print("Test MSE:", test_loss)
print("Test MAE:", test_mae)
```

### 3. Указываем путь к набору данных:

```python
imagePaths = sorted(list(paths.list_images("/content")))
random.shuffle(imagePaths)
print("Всего изображений:", len(imagePaths))
```

### 4. Загружаем набор данных и формируем общую выборку:

```python
def load_images(imagePaths, compression_size = 32):
  data = []
  labels = []
  for imagePath in imagePaths:
    image = cv2.imread(imagePath)
    image = cv2.resize(image, (compression_size, compression_size)).flatten()

    data.append(image)
    label = imagePath.split(os.path.sep)[-2]
    labels.append(label)

  return data, labels

compression_size = 32
data, labels = load_images(imagePaths, compression_size)
```

#### 4.1. Нормализуем данные:

Масштабируем интенсивности пикселей в диапазон [0, 1], а также преобразуем данные в массивы NumPy для повышения производительности


```python
data = np.array(data, dtype="float") / 255.0
labels = np.array(labels)
```

#### 4.2. Формируем обучающую и валидационную выборки:

Разделяем данные на обучающую и валидационную выборки, используя 75% данных для обучения и оставшиеся 25% для валидации

```python
(X_train, X_val, Y_train, Y_val) = train_test_split(data, labels, test_size=0.25)
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
lb = LabelBinarizer()
Y_train = lb.fit_transform(Y_train)
Y_val = lb.transform(Y_val)
```

### 5. Создание модели:

#### 5.1. Определим архитектуру модели с помощью Keras:

```python
input_shape = (compression_size * compression_size * 3,)
input_shape
```

```python
model = Sequential([
    Dense(1024, input_shape=input_shape, activation="relu"),
    Dense(128, activation="relu"),
    Dense(len(lb.classes_), activation="softmax")
])

model.summary()
```

#### 5.2. Установим значение параметров скорости обучения и общего числа эпох


```python
learning_rate = 0.001
epochs = 300
batch_size = 32

opt = Adam(learning_rate=learning_rate)
```

#### 5.3. Компилируем модель:

```python
model.compile(
    loss="categorical_crossentropy",
    optimizer=opt,
    metrics=["accuracy"]
)
```

### 6. Обучение модели искусственной нейронной сети:

#### 6.1. Добавим колбэки для сохранения лучшей модели по валидационной точности и остановки обучения, в случае, если модель перестала обучаться:

```python
from tensorflow.keras.callbacks import ModelCheckpoint, EarlyStopping
```

```python
checkpoint = ModelCheckpoint(
    "best_fruit_model.keras",
    monitor="val_accuracy",
    save_best_only=True,
    mode="max",
    verbose=1
)

earlystop = EarlyStopping(
    monitor="val_accuracy",
    patience=5,
    restore_best_weights=True,
    mode="max",
    verbose=1
)

callbacks = [checkpoint, earlystop]
```

#### 6.2. Запустим процесс обучения модели:

```python
history = model.fit(
    X_train, Y_train,
    validation_data=(X_val, Y_val),
    batch_size=batch_size,
    epochs=epochs,
    callbacks=callbacks,
    verbose=1
)
```

#### 6.3. Загружаем лучшую модель:

#### 9.2. Реализуем функцию для предобработки входных данных в той же последовательности, что и при формировании обучающей выборки:

```python
model = load_model("fruit_classifier.keras")

with open("fruit_labels.pkl", "rb") as f:
    lb_loaded = pickle.load(f)

def preprocess_image(image, compression_size=32):
    image = cv2.resize(image, (compression_size, compression_size))
    image = image.astype("float32") / 255.0
    image = image.flatten()
    image = np.expand_dims(image, axis=0)
    return image
```

#### 9.3. Определим функцию для осуществления прогноза по загруженной модели:

```python
def predict_fruit(image):
    image = cv2.cvtColor(image, cv2.COLOR_RGB2BGR)
    processed = preprocess_image(image, compression_size=compression_size)

    probs = model.predict(processed, verbose=0)[0]
    pred_idx = np.argmax(probs)
    pred_label = lb_loaded.classes_[pred_idx]

    result = {lb_loaded.classes_[i]: float(probs[i]) for i in range(len(lb_loaded.classes_))}
    return result
```
