# **Задание 1. Определение концептуальных основ дипломного проекта в области информационных систем**



В рамках дисциплины "Управление проектами информационных систем" представьте развернутое концептуальное обоснование вашего проекта, ответив на следующие вопросы:


#### **1. Обоснование необходимости разработки**


Для кого предназначено разрабатываемое Вами решение и какие потребности целевой аудитории оно призвано удовлетворить?




*Обоснуйте актуальность решаемой проблемы для конкретных заинтересованных сторон.*

Решение предназначено для преподавателей, студентов, ведущих мероприятий и создателей образовательного контента. Их основная потребность - быстро и без ручной рутины получать качественные квизы на основе собственных материалов. Сервис решает проблему долгой подготовки вопросов: пользователь загружает свои документы, система автоматически анализирует содержание и формирует структурированный набор вопросов, который можно сразу использовать в учебном процессе или мероприятии.

---

#### **2. Ожидаемые результаты проекта**



Опишите конкретные результаты, которые планируется получить по завершении проекта. Какой программный продукт, методика или информационная система будет создан/создана в результате вашей работы?

По завершении проекта будет создан веб-сервис, включающий интерфейс для загрузки материалов, модуль обработки текстов, систему генерации вопросов на базе локальной LLM и механизм экспорта готового квиза в формат PowerPoint и JSON. Это полноценный инструмент автоматизированной подготовки учебных и презентационных материалов.

---

#### **3. Методология использования**


Каким образом предполагается практическое применение разрабатываемого решения? Опишите основные сценарии взаимодействия пользователей с разрабатываемым решением.

Практическое применение заключается в том, что пользователь входит в систему, загружает свои тексты, получает автоматически сгенерированный квиз, при необходимости редактирует его и скачивает итоговую презентацию. Вся работа сведена к нескольким шагам, а итоговый файл можно сразу вывести на экран во время занятия, лекции или интерактивной игры.

---

#### **4. Входные данные и их источники**


Какие типы данных будут использоваться в качестве входных данных для вашего решения? В каком формате они будут представлены и из каких источников планируется их получение?

Входными данными являются текстовые материалы пользователя: PDF, DOCX, TXT и HTML файлы. Они содержат исходный контент, из которого формируются вопросы. Все файлы проходят извлечение текста и разбиение на фрагменты, чтобы модель могла корректно анализировать структуру и смысл документа.

---

#### **5. Выходные данные и практическая значимость**


Какие результаты будет формировать Ваше решение на выходе?

На выходе система формирует готовый квиз в виде презентации PowerPoint с вопросами и ответами, а также структурированный JSON-файл, который может использоваться для хранения или дальнейшей обработки. Пользователь получает законченный продукт, не требующий дополнительной подготовки.

Практическая ценность решения в том, что оно экономит время и снижает нагрузку на преподавателей и ведущих, позволяя за минуты подготовить материалы, на которые в обычных условиях уходят часы. Пользователь получает качественный, логичный и оформленный квиз, который сразу готов к применению, что резко повышает эффективность подготовки и проведения образовательных и интерактивных мероприятий.
