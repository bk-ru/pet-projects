# **Задание 1. Определение концептуальных основ дипломного проекта в области информационных систем**

#### **1. Обоснование необходимости разработки**


Основная целевая аудитория будет: преподаватели и учителя образовательных учреждений; HR-специалисты; Создатели образовательного контента. Их функционал будет включать: создание квизов и тестов на основе учебных материалов — трудоемкий процесс, требующий временных затрат на анализ документов, выделение ключевых понятий и формулировку вопросов с правдоподобными терминами.


Существующие решения требуют ручного ввода всех вопросов, либо надо генерировать в чат - нейронках. Решение закрывает разрыв между ручным созданием квизов и автоматической генерацией.

#### **2. Ожидаемые результаты проекта**



**Веб-платформа для автоматизированного создания квизов, в которой будет системас регистрацией и личными кабинетами, модуль обработки документов с извлечением текстовой информации из различных форматов, а также генерация правдоподобных неверных вариантов ответов, конструктор квизов с возможностью редактирования автоматически сгенерированных данных, система экспорта готовых квизов в форматах типа JSON, система подключения других учащихся по Qr коду**

---

#### **3. Методология использования**


Преподаватель:
1. Регистрация/вход в систему
2. Загрузка учебного документа (лекция, статья, глава учебника)
3. Автоматический анализ документа и выделение ключевых концепций
4. Генерация вопросов с вариантами ответов
5. Редактирование сгенерированных вопросов
6. Настройка параметров квиза (время, проходной балл)
7. Публикация или экспорт квиза
8. Вывод QR-кода для подключения студентов/ученовек/других

Студент:
1. Загрузка конспекта или статьи для подготовки к экзамену
2. Автоматическая генерация тренировочного квиза
3. Прохождение теста с получением результатов
4. Анализ ошибок с отсылками к исходному тексту

Руководитель / ответственный за инструктажи
1. Загрузка регламентов, инструкций, политик компании
2. Создание проверочных тестов для сотрудников

---

#### **4. Входные данные и их источники**


**Типы входных данных:**

**Текстовые документы:**
   - Формат: PDF, DOCX, TXT, JSON
   - Содержание: учебные материалы, научные статьи, доклады, технические документации

**Пользовательские параметры:**
   - Количество вопросов
   - Тип вопросов (одиночный/множественный выбор)
   - Количество вариантов ответа

**Источники данных:**
- Локальные файлы пользователя (прямая загрузка через веб-форму)
- База загруженных материалов

---

#### **5. Выходные данные и практическая значимость**


**Выходные данные:**

   - Вопросы с выделенными ключевыми параметрами
   - n вариантов ответа на каждый вопрос
   - Правильные ответы


   - JSON

**Практическая ценность для заинтересованных сторон:**

**Для всех:**
- Во время игры, лучше всего запоминается информация

**Для преподавателей:**
- Сокращение времени подготовки тестов
- Возможность организации командной работы учащихся

**Для студентов:**
- Инструмент самопроверки знаний без помощи преподавателя
- Подготовка к экзаменам

**Для корпоративного сектора:**
- Быстрое создание проверочных тестов при обновлении регламентов
- Стандартизация процесса аттестации сотрудников

**Для образовательных учреждений:**
- Формирование банка тестовых заданий с минимальными трудозатратами
- Интересное решение для дистанционного обучения
