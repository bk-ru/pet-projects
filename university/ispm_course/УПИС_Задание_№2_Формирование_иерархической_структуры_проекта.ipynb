# **Задание 2. Формирование иерархической структуры проекта в контексте управления проектами информационных систем**






#### **Пример №2. Специализированная структура IT-проекта**

```
1. ОБЗОР ПРЕДМЕТНОЙ ОБЛАСТИ И СУЩЕСТВУЮЩИХ РЕШЕНИЙ
   1.1. Исследование текущего состояния предметной области
      1.1.1. Современные тенденции и проблемы в исследуемой сфере
      1.1.2. Существующие подходы к решению выявленных проблем

   1.2. Обзор существующих систем и приложений
      1.2.1. Отечественные решения и их функциональные возможности
      1.2.2. Зарубежные аналоги и опыт их применения

   1.3. Обзор технологий и инструментов разработки
      1.3.1. Современные языки программирования и фреймворки
      1.3.2. Платформы и среды разработки
      1.3.3. Подходы к созданию аналогичных решений (типовые стеки разработки)

2. АНАЛИЗ И ФОРМИРОВАНИЕ ТРЕБОВАНИЙ К СИСТЕМЕ
   2.1. Анализ недостатков существующих решений
      2.1.1. Выявление функциональных ограничений аналогов
      2.1.2. Проблемы производительности, масштабируемости, ограниченной области применения

   2.2. Выбор технологического стека на основе анализа
      2.2.1. Обоснование выбора языков программирования
      2.2.2. Определение архитектуры и платформы развёртывания
      2.2.3. Выбор инструментов разработки и тестирования

   2.3. Формирование требований к разрабатываемой системе
      2.3.1. Функциональные требования и пользовательские сценарии (диаграмма вариантов использования, диаграмма деятельности)
      2.3.2. Нефункциональные требования
      2.3.3. Критерии оценки эффективности решения (индивидуально)

3. ПРОЕКТИРОВАНИЕ И РЕАЛИЗАЦИЯ СИСТЕМЫ
   3.1. Проектирование архитектуры приложения
      3.1.1. Проектирование структуры базы данных (ER-диаграмма)
      3.1.2. Проектирование модулей и их взаимосвязей (Диаграмма классов)
      3.1.3. UML-диаграммы и схемы взаимодействия компонентов (Диаграмма компонентов, диаграмма автоматов, диаграмма деятельности и т.д.)

   3.2. Программная реализация системы
      3.2.1. Разработка серверной части и бизнес-логики
      3.2.2. Создание пользовательского интерфейса
      3.2.3. Интеграция с внешними системами и API

   3.3. Внедрение и тестирование
      3.3.1. Развёртывание системы в рабочем окружении
      3.3.2. Функциональное и нагрузочное тестирование (индивидуально)

```

# Содержание:

```
1. Анализ предметной области и технологий автоматизированной генерации квизов
   1.1. Современное состояние области подготовки учебных материалов
      1.1.1. Проблемы ручного создания квизов и высокая трудоёмкость процесса
      1.1.2. Применение искусственного интеллекта для автоматизации образовательных задач
      1.1.3. Ограничения существующих подходов к обработке текстовых документов

   1.2. Анализ существующих сервисов генерации тестов
      1.2.1. Функциональные особенности популярных отечественных решений
      1.2.2. Возможности зарубежных платформ и их применимость для русскоязычных материалов
      1.2.3. Причины отсутствия полноценных систем, работающих с локальными LLM

   1.3. Технологические основы разработки системы
      1.3.1. Использование локальных языковых моделей (Llama) для анализа документов
      1.3.2. Принцип работы RAG и его роль в извлечении контекста
      1.3.3. Технологии генерации презентаций и структура PPTX-файлов

2. Формирование требований и обоснование выбора технологий
   2.1. Выявление недостатков существующих решений
      2.1.1. Функциональные ограничения доступных сервисов
      2.1.2. Недостаточная адаптация под русскоязычные документы и материалы
      2.1.3. Отсутствие приватной обработки данных и поддержки локальных моделей

   2.2. Обоснование технологического стека
      2.2.1. Выбор FastAPI, Python и Vue для реализации сервиса
      2.2.2. Использование PostgreSQL и pgvector для хранения данных и семантического поиска
      2.2.3. Выбор Llama и GPU-инфраструктуры для генерации вопросов

   2.3. Формирование требований к разрабатываемому веб-сервису
      2.3.1. Функциональные требования на основе пользовательских сценариев
      2.3.2. Нефункциональные требования: масштабируемость, приватность, производительность
      2.3.3. Критерии оценки качества генерации квизов и удобства использования

3. Проектирование и реализация системы генерации квизов
   3.1. Архитектура системы
      3.1.1. Проектирование структуры базы данных для хранения пользователей, документов и квизов
      3.1.2. Архитектура серверных модулей: загрузка, парсинг, генерация, экспорт
      3.1.3. Схемы взаимодействия фронтенда, бекенда и модуля LLM

   3.2. Реализация ключевых модулей сервиса
      3.2.1. Модуль обработки документов и формирования чанков для RAG
      3.2.2. Механизм генерации вопросов и редактор квиза
      3.2.3. Модуль экспорта итогового квиза в формат PowerPoint

   3.3. Развёртывание и тестирование системы
      3.3.1. Развёртывание сервисов на локальной и облачной GPU-инфраструктуре
      3.3.2. Функциональное тестирование модулей обработки, генерации и экспорта
      3.3.3. Оценка качества решения и направления дальнейшего развития
```
